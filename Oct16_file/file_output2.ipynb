```python
# 파일 복사
# 기존 파일의 내용을 읽어와서 
# 새로운 파일을 만들고
# 읽어온 내용을 출력한다

# dream.txt문서를 복사한 new_dream.txt문서 생성

# open
with open('dream.txt', 'r') as f:
    contents = f.read()

# open
with open('new_dream.txt','w') as f:
    f.write(contents)

print('end')
```

```python
#하나의 파일에 대해 읽기 쓰기 겸용
with open('new_dream.txt', 'r+') as f:
    d_data = f.read()
    f.write('\n\n\n')
    f.write(d_data)

print('end')
```
