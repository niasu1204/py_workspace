```python
# 파일 내용 읽어오기
# open('파일경로')
# read()
# 경로 나눠주는 \ -> 이스케이프 문자열로 인식(\n, \t) -> 없는 문자 -> 에러
# 문자열 앞에 r을 붙여주면 단순 문자열로 인식

# 절대 경로 - 처음부터 모든 경로 다 지정
f = open(r'C:\Users\soldesk\py_workspace\Oct16_file\dream.txt')
text = f.read() # 파일 내용을 하나의 문자열로 가져옴
f.close() # 연결 종료

print(text)
```

```python
# 현재 작업 위치를 기준으로 경로 지정
# 같은 경로에 있다면 파일 이름만 사용
# 상대 경로 - 작업 위치를 기준으로 경로 지정
f = open(r'dream.txt')
text = f.read() # 파일 내용을 하나의 문자열로 가져옴
f.close() # 연결 종료

print(text)
```

```python
f = open(r'dream.txt')
text2 = f.readline() # 줄바꿈, \n를 기준으로 한줄 읽어옴
text3 = f.readline()
f.close() # 연결 종료

print(text2)
print(text3)
```

```python
f = open(r'dream.txt')
t_list = f.readlines()
f.close()

print(t_list)
for t in t_list:
    print(t)
```
