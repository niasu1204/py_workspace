```python
# 대입 연산자 = 문장 내 우선순위 제일 마지막!!
a = 10 # 오른쪽의 값을 왼쪽 변수에 저장한다.(대입, 할당)            

a1 = a - 3
```

```python
# 복합 대입 연산자 += -= *= /= //= %= **=
a + 5 # 더한 결과를 저장하지 않으면 변수의 값은 바뀌지 않음
print(a)

a += 5 # 왼쪽 변수의 값과 오른쪽 식의 결과값을 더하고 그 값을 왼쪽 변수에 저장한다.
print(a)

a = a + 5 
```

```python
a -= 10 # 1. 왼쪽 변수 값 - 오른쪽 값 2. 결과물 왼쪽 변수에 저장
print(a)
```

```python
b = 2
b **= 5  # 제곱 구하기
print(b)
```

```python
# 비교 연산자 : 숫자라면 크기 비교, 문자열 순서(사전 순서) 비교
# 결과는 항상 True/False로 나온다

n1 = 10
n2 = 30

print(n1 < n2)
print(n1 > n2)
print(n1 <= n2)
print(n1 >= n2)
```

```python
# 문자열 비교시 같은 인덱스(위치)에 있는 한글자씩 비교
# 공백 > 숫자 > 알파벳 대문자 > 소문자 > 한글 순으로 비교

s1 = '10'
s2 = '8'

print(s1 < s2)
```

```python
s3 = 'AAb'
s4 = 'AAC'
print(s3 < s4) # 'b' < 'C'
```

```python
s5 = 'AA'
s6 = 'AA'
print(s5 < s6)
```

```python
# 타입이 다르면 비교 X

s1 = '10'
s2 = 8

print(s1 < s2)
```

```python
# == 값이 일치하면 True
# != 값이 일치하지 않으면 True(다르다)

s7 = 'A BC'
s8 = 'A BC'
print(s7 == s8)
print(s7 != s8)
print(s7 is s8) # 주소값이 일치하면 True(값은 상관 X)


s9 = 10
s10 = '10'
print(s9 == s10) # 타입이 다르면 다른 값
```

```python
# is 연산자 : 변수가 가지고 있는 메모리 주소값 비교
# None 값을 비교할때 관례적으로 사용
# None : 객체 없음 -> 주소값이 없음
value = None
print(value is None)
print(value == None)
```
