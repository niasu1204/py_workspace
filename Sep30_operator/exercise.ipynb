```python
'''
* 문제를 잘 쪼개야 한다!!
1. 입력 / 저장 -> 저장해두어야 할 값
2. 연산(처리)
3. 결과 출력
'''
```

```python
'''
   1. 복리 이자 계산기
   초기 원금(principal): 1000000원 저장
   연이율(rate): 5%(0.05)
   -1년 후 총 금액을 계산(원금 = 원금 + 원금 * 이율)하여 초기원금에 다시 저장
   -2년 후 총 금액을 동일한 방식으로 다시 계산
   -2년 후 최종 금액을 출력하되 정수만 출력하세요
'''
principal = 1000000
rate = 0.05

# 1년 후
# principal = principal + principal * rate
principal += principal * rate

# 2년후 
principal += principal * rate

print(f'2년후 최종 금액: {int(principal):,}원')
```

```python
'''
    임의의 정수가 주어졌을때 조건에 맞으면 True, 아니면 False를 출력하는 조건식을 작성하세요
    num = 20

    1. 짝수인지 확인하는 조건식
    출력 결과 : 짝수 : True

    2. 10보다 크면서 3의 배수인지 확인하는 조건식
    출력 결과 : 10보다 크면서 3의 배수인가? : False
'''
num = int(input('정수를 입력하세요'))
# 1
is_even = num % 2 == 0 
print(f'짝수 : {is_even}')

# 2
print(f'10보다 크면서 3의 배수인가? : { num > 10 and (num % 3 == 0)}')
```

```python
'''
    3가지 조건 중 하나라도 충족시 할인 대상
    1. 나이가 13세 미만이거나 65세 이상인 경우 
    2. vip회원인 경우(vip == true)
    3. 보유한 쿠폰 코드(coupon_code)에 '2026'이라는 문자열이 포함된 경우

    출력 결과 : 할인 대상 : True/False 
'''
age = 19
vip = False  # True : vip회원, False : vip 아님
coupon_code = 'Oct_2025'

is_discount = (age < 13 or age >= 65) or vip or '2026' in coupon_code
print(f'할인 대상 : {is_discount}')


```
