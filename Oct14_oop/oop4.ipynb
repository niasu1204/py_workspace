```python
# 상속 : 클래스가 클래스의 멤버를 물려받음
'''
Dog - name
    - eat
    - sound
'''
class Dog:
    def __init__(self, name):
        self.name = name
    def eat(self):
        print('밥 먹음')
    def sount(self):
        print('멍멍')
'''
Person
    - name
    - eat
    - sound
    - study
'''

class Person:
    def __init__(self, name):
        self.name = name
    def eat(self):
        print('밥 먹음')
    def sount(self):
        print('말로 해')
    def study(self):
        print('공부는 평생...')    
```

```python
#여러 클래스 간의 공통 멤버 -> 부모 클래스 선언
# 공통 멤버는 물려받고 다른 멤버는 직접 선언

# 부모 클래스
class Mammal:
    def __init__(self, name):
        self.name = name
    def eat(self):
        print('밥 먹음')
    def sound(self):
        print('소리가 남')

class Dog(Mammal): # 상속. init, eat, sound 다 상속받음
    pass

class Person(Mammal):
    def study(self):
        print('공부는 평생..')

d1 = Dog('바둑이')
d1.eat()
d1.sound()

p1 = Person('김철수')
p1.eat() # 물려받은 멤버
p1.study() # 직접 선언한 멤버
```

```python
'''* 두 클래스의 공통점을 찾아 super클래스인 Person을 선언하세요

직원( Employee)
-이름(name), 나이(age), 성별(gender), 부서(dept), 연봉(sal)
-자기 소개 (about_me): 이름, 나이, 성별 출력
-일하기(do_work) : print('일을 열심히 합니다.')

학생(student)
-이름(name), 나이(age), 성별(gender), 전공(major), 성적(grade)
-자기 소개 (about_me): 이름, 나이, 성별 출력
-성적 공개 (show_info): 전공, 이름, 성적 출력
'''
```

```python
class Person:
    def __init__(self, n, a, g):
        self.name = n
        self.age = a
        self.gender = g

    def about_me(self):
        print(f'name:{self.name}, age:{self.age}, gender:{self.gender}')    


class Employee(Person):
    pass
class Student(Person):
    pass        
```
