```python
class Appliance:
    def __init__(self, b):
        self.brand = b
        self.__power = False # 비공개 속성(False -> 전원 꺼짐)

    def turn_on_off(self):
        self.__power = not self.__power
        if self.__power :
            print(f'{self.brand} 전원이 켜졌습니다.')
        else: print('전원이 꺼졌습니다.')        
    # getter : 변수의 값을 가져오는(반환하는) 함수
    def is_power_on(self):
        return self.__power

class TV(Appliance):
    def __init__(self, b, c = 1):
        super().__init__(b)
        self.channel = c
    def set_channel(self, c):
        if self.is_power_on(): # 물려받아서 인스턴스(self)가 가지고 있음
            self.channel = c
            print(f'{self.channel}번 채널로 변경되었습니다.')
        else:
            print('전원을 먼저 켜주세요')    

# --- 테스트 코드 ---
my_tv = TV("삼성")
my_tv.set_channel(7)
my_tv.turn_on_off()
my_tv.set_channel(7)

my_tv.turn_on_off()
my_tv.set_channel(11) 
```
