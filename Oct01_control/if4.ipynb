```python
# if 축약식(한줄로 표현)
# 변수 = 참일때 저장할 값  if 조건식 else 거짓일때 값
# 조건식 수행 - True if 앞의 식 수행 후 변수에 저장
#  False - else 뒤에 있는 식이 수행
x = 9
y = '짝수' if x % 2 == 0 else '홀수'
print(y)
```

```python
num = int(input('정수 넣어라!!'))
# if num % 4 == 0 : 
#     print('4의 배수입니다.')
# else:
#     print('4의 배수가 아닙니다.') 

result = '4의 배수입니다.' if num % 4 == 0 else '4의 배수가 아닙니다.'
print(result)
```

```python
# 두 사람이 주사위를 던져 더 큰 수가 나오면 이기는 게임
# A가 이기거나 B가 이기거나 비기는 결과가 나오도록 프로그램 작성
# 1 ~ 6 사이의 값 하나 -> 주사위 

import random
print(random.randrange(1,7)) # 1~6사이의 정수 랜덤값 발생
```

```python
import random
a = random.randrange(1,7)
b = random.randrange(1,7)
print(a, b)
# if
if a > b:
    print('A가 이겼습니다.')
elif a < b:
    print('B가 이겼습니다.')
elif a == b:
    print('비겼습니다.')        

# 축약형
result = 'A win' if a > b else ('B win' if a < b else 'A, B same' )
print(result)
```
