```python
# 조건(True/False)에 따라 코드를 수행하는 조건문(conditional Statement)
'''
if 조건식:
    조건식의 결과가 True일때 수행할 문장1
    조건식의 결과가 True일때 수행할 문장2
    조건식의 결과가 True일때 수행할 문장3
            
if문 상관없이 실행할 문장
    
* 코드블록의 범위를 결정 -> 들여쓰기(tab키)
'''
```

```python
if True:
    print('수행1')
    print('수행2')

print('if 상관없는 문장')    
```

```python
if '': # bool타입의 값으로 자동 변환
    print('수행')
```

```python
age = 10
if age <= 19:
    print('미성년자입니다.')
```
