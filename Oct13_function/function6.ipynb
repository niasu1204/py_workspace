```python
#전역변수(grobal)/지역변수(local)

# 선언 위치에 따라 유효 범위(scope)달라짐
# 전역변수 : 함수 외부에서 선언한 변수. 어디서든 사용 가능(함수 안, 밖)
# 지역변수 : 함수 내부에 선언한 변수. 
            # 함수가 호출 -> 실행 -> 지역변수가 메모리에 올라감
            # -> 함수 종료 -> 메모리 반납(지역변수 소멸)
            # 함수 안에서만 사용 가능
            # 매개변수도 지역변수 
```

```python
# 변수 검색 순서 (LEGB규칙)
# 1. 함수 내부(Local)
# 2. 바깥쪽 함수(Enclosing)
# 3. 전역변수(파일)(Global)
# 4. 파이썬 기본 내장 함수/변수(len(), print()...) (Built-in)
```

```python
# 전역변수와 지역변수가 이름이 같다면 -> 함수 안에서는 지역변수가 우선권을 가짐
def f(x): # 지역변수
    y = x
    x = 5 # 지역변수 x의 값 변경.
    return y * y

x = 3 # 전역변수
print('f(x):',f(x))
print('x:',x) # 전역변수 x는 값이 바뀌지 않음
```

```python
s = 'I love Paris.'
def f1():
    s = 'I love London.' # 지역 변수 선언
    print('f1 안에서 s:', s) # 지역 변수 사용

f1()
print('f1()끝난 후: ', s)

def f2():
    print('f2 안에서 s:', s) # 전역 변수 사용

f2()
```

```python
# 변수에는 항상 주소값 저장 - 매개변수에도 주소값이 전달됨
def spam(eggs:list): # 
    eggs.append(1) # 주소값이 전달-> 리스트 접근해서 작업 가능 
    eggs = [2,3] # 새로운 리스트 객체 생성 후 지역변수에 저장
    print('eggs:',eggs)


ham = [0]
spam(ham)
print('ham:', ham)    
```
