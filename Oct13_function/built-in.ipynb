```python
# 빌트인(built-in) 함수

print('Hello') # 출력
input('메세지:') # 키보드로 값 입력
type(10) # 값의 자료형 확인
len('Python') # 길이 확인
int('100') # 정수 변환
float('3.14') # 실수 변환
str(100) # 문자열로 변환
bool(0) # True/False로 변환
range(1,4) # 정수 범위 생성
```

```python
# 함수 설명
help(len)
```

```python
# 숫자 계산 관련 함수
scores = [80, 90, 100]
print(sum(scores))
print(max(scores), min(scores))

print(abs(-15)) # 절대값
print(round(3.14159, 3)) # 지정된 자리에서 반올림
```

```python
# 리스트와 반복문에서 자주 사용되는 함수
nums = [30,10,20]
print(sorted(nums))
print(sorted(nums, reverse=True))


print(list(reversed(nums)))

#인덱스와 값을 한번에 얻기
names = ['김','이','박, 강']
for i, name in enumerate(names):
    print(i, name)
```

```python
# zip() - 여러 자료에 있는 값들을 같은 인덱스끼리 묶기
names = ['scott','james','miller']
scores = [90, 80, 100]

for name, score in zip(names, scores):
    print(name, score)

for t in zip(names, scores):
    print(t)    
```

```python
a = [1,2,3,4]
b = [10, 20, 30]

# 같은 인덱스에 값이 없으면 안가져옴
for i in zip(a, b):
    print(i)
```

```python
for i, (a,b) in enumerate(zip(a,b)):
    print(f'{i}인덱스: {a}:{b}')
```

```python
a = [1,2,3,4]
b = [10, 20, 30]
c = [11, 22, 33]
for t in zip(a,b,c):
    print(t)
```

```python
# all() : 모든 값이 True 일때만 True
# any() : 하나라도 True면 True
a = [True, True, True, True]
b = [True, True, True, False]

print(all(a))
print(all(b))
print(any(a))
print(any(b))
```

```python
# 리스트에 저장된 값이 전부 양수인지 확인
nums = [2, 4, 7, 8]
print(all([ n > 0 for n in nums ]))
if  all([ n > 0 for n in nums ]):
    print('전부 양수 맞음')
else:
    print('양수가 아닌 값도 있음')    
```
