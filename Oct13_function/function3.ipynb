```python
# 매개변수(parameter) : 전달받은 값 저장 공간
# 인자(argument) : 작업할때 전달받는 값
```

```python
# 키워드 인자
# 함수 호출시에 argument 전달 -> 저장할 매개변수 이름을 지정해서 전달
# 이름(name)과 직책(position)을 전달받아 인사말을 출력하는 greet 함수

def greet(name, position):
    print(f'{name} {position}님 안녕하세요')

greet('김솔데', '부장')
greet(position='팀장', name='김스캇') # 매개변수 이름에 맞춰서 저장
# print('출력 내용', end='', sep='/')

# 주의할 점 - 키워드 인자는 뒤에 명시
# greet(name='김제임스', '프로')
# greet('프로', name='김제임스') # 키워드가 없는 값은 무조건 순서대로 들어감
                                # name에 값이 여러개 들어가는 상황 -> error
greet('김제임스', position='pro')                                
```
