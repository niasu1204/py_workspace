```python
#1. int값 한개 받아서 짝수라면 True, 
# 아니라면 False를 반환하는 is_even함수를 선언하고 호출하세요

def is_even(n:int) -> bool:
    return n % 2 == 0

print(f'is_even(2): {is_even(3)}')
```

```python
# 2. 문자열 2개 받아서 길이를 비교하는 함수 longer_string함수를 선언하세요
# 길이가 긴 문자열을 리턴하고 만약 길이가 일치한다면 'same'이라는 문자열을 리턴하세요
# 빈 문자열이 입력된 경우 계산 안하고 바로 글자가 있는 문자열을 리턴하세요
# 둘 다 빈문자열(falsy)인 경우 'wrong' 리턴하세요


def longer_string(str1:str, str2:str) -> str:
    # 둘 다 빈문자열인 경우
    if not str1 and not str2:
        return 'wrong'
    # 하나가 빈문자열인 경우
    # if not str1:
    #     return str2
    # if not str2:
    #     return str1
    if not str1 or not str2:
        return str1 or str2  # 첫번째 문자열이 값이 있으면(True) 그 값 리턴
                             # 없다면(False) 뒤에 있는 값 리턴

    # 길이가 긴 문자열을 리턴, 아니면 'same'리턴 
    if len(str1) > len(str2):
        return str1
    elif len(str1) < len(str2):
        return str2
    else : return 'same'

print(longer_string('',''))
print(longer_string('python',''))
print(longer_string('','hello'))
print(longer_string('python','hello'))
print(longer_string('pytho','hello'))
```

```python
# 3. int만 저장된 list를 받아서 평균값을 계산해 리턴하세요
# 빈 리스트(falsy)는 계산하지 않고 0을 리턴합니다.
def average(nums:list[int]) -> float:
    if not nums:
        return 0
    return sum(nums) / len(nums)

print(average([]))
print(average([5,4,5,2,5]))
```

```python
# 4. 학생의 이름과 성적을 받아 가장 잘 한 학생의 이름을 리턴하는 best_student함수를 선언하세요
# {"kim": 80, "lee": 95, "park": 90}

def best_student(scores:dict[str:int]) -> str:
    return max(scores, key=scores.get) # key: 비교 기준

print(best_student({"kim": 80, "lee": 95, "park": 90})) 


d1 = {"kim": 80, "lee": 95, "park": 90}

# max(딕셔너리) ->  key를 기준으로 비교
# 'kim', 'lee', 'park' -> 'park' 키 값 반환
print(max(d1))

# max(딕셔너리, key=비교 대상을 가져오는 함수)
# 딕셔너리에 get함수 사용해 value 가져와서 최대값 찾음
# 최대값의 key 반환
print(max(d1,  key= d1.get ))


```

```python
# 5. 비밀번호 검증하는 함수 validate_password함수를 선언하세요
# 문자열로 비번을 받아와
# 글자수가 8자 이상, 숫자, !가 각각 하나라도 있어야 한다
# 조건을 통과하면 True, 아니면 False 리턴

def validate_password(password:str) -> bool:
    # 글자수가 8자 이상
    if len(password) < 8:
        return False
    # 숫자, !가 각각 하나라도 있어야 한다
    has_d = False  # 숫자가 하나라도 있으면 True
    has_special = False # 느낌표가 하나라도 있으면 True

    for ch in password:
        if ch.isdigit(): # 글자가 숫자로 이루어져 있으면
            has_d = True
        if ch == '!':
            has_special = True

    return has_d and has_special            

pwd = input('사용할 비밀번호를 입력하세요')
if validate_password(pwd):
    print('입력한 비번은 사용할 수 있습니다.')
else:
    print('입력한 비번은 사용할 수 없습니다.')    


def validate_password2(password:str) -> bool:
    is_valid = all( [ len(password) >= 8 , 
                    '!' in password, 
                    any(ch.isdigit() for ch in password) 
                        ])
    return is_valid

```
