```python
# 부모 클래스
class Mammal:
    def __init__(self, name):
        self.name = name
    def eat(self):
        print('밥 먹음')
    def sound(self):
        print('소리가 남')

class Dog(Mammal): # 상속. init, eat, sound 다 상속받음
    # 상속받은 함수를 현재 클래스에 맞게 재정의 -> 오버라이딩
    # 이름 변경 X, 수행부분만 변경
    def sound(self):
        print('멍멍')


class Person(Mammal):
    def sound(self):
        print('말을 한다')
    def study(self):
        print('공부는 평생..')

d1 = Dog('바둑이')
d1.eat()
d1.sound() # 재정의된 함수 실행

p1 = Person('김철수')
p1.eat() # 물려받은 멤버
p1.study() # 직접 선언한 멤버
p1.sound() # 재정의된 함수 실행
```

```python
class Person:
    def __init__(self, n, a, g):
        self.name = n
        self.age = a
        self.gender = g

    def about_me(self):
        print(f'name:{self.name}, age:{self.age}, gender:{self.gender}')    

#부서(dept), 연봉(sal)
class Employee(Person):
    # 현재 클래스에 맞게 다시 선언
    def __init__(self, n, a, g, d, s):
        super().__init__(n, a, g) # super() 부모 클래스. | 부모에 선언된 멤버는 부모클래스에서 처리
        self.dept = d
        self.sal = s

    def do_work(self):
        print('일을 열심히 합니다.')

#  전공(major), 성적(grade)
# 성적 공개 (show_info): 전공, 이름, 성적 출력
class Student(Person):
    def __init__(self, n, a, g, m, gr):
        super().__init__(n,a,g)
        self.major = m
        self.grade = gr

    def show_info(self):
        print(f'major:{self.major}, name:{self.name}, grade:{self.grade}')    


# 'kim'/ 30세/ '남'/'개발팀'/ 5000
kim = Employee('kim', 30, '남', '개발팀', 5000)
kim.about_me()
kim.do_work()

# 'smith'/ 20세/ '남' /'철학'/ 'A'
smith = Student('smith',20, '남', '철학', 'A')
smith.about_me()
smith.show_info()
```
