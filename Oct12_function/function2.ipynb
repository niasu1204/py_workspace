```python
# return 

def f():
    return 1

print(f())# 리턴값이 있다면 리턴값 들고와서 사용
```

```python
def h():
    print('함수 안에서 수행')

# 리턴문이 없는 함수 결과 출력 -> None
print(h())    
```

```python
def h2(x:int):
    if x == 10:
        return # 특정 조건(if)시 함수를 바로 끝내고 싶다
               # return만 명시
               # 더이상 함수 수행 안하고 값없이 호출한 지점으로 돌아간다. 
    else:
        print('10 아님')
    print(x)

h2(10)
print('-'*10)
h2(11)    
```

```python
# return 값, 값, 값
# return에 값이 여러개라면?
# 실제 반환 값은 1개 -> tuple로 묶어서(패킹) 반환

def h3():
    return 10, 20, 30
print(h3(), type(h3()))
a, b, c = h3() # 언패킹(개수 일치해야 함!!)
print(f'{a}, {b}, {c}')
# 전달받을 값의 개수를 모른다면?
a1, *b1 = h3()  # *마지막 변수 : 남은 값들을 모아서 패킹
print(f'{a1}, b1: {b1}')


```
