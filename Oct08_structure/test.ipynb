```python
# 1. 정수로 주어진 리스트에서
# 중복을 제거한 뒤 오름차순으로 정렬한 리스트 만드세요
nums = [4,2,2,1,3,4]

# 1. 중복을 제거한 뒤
s = set(nums)
# 2. 오름차순으로 정렬
print(sorted(s))

# 한줄로
print(sorted(set(nums)))
```

```python
# 2. 두개의 리스트가 주어질때 공통된 요소만 가지고 있는 리스트로 만들어주세요
a = [1,2,3,4]
b = [3,4,5,6]

# 공통된 요소만 -> 교집합
set_a = set(a)
set_b = set(b)
set_ab = set_a & set_b
print(set_ab)
#리스트로 만들
print(list(set_ab))

#한줄로 처리
print(list(set(a)&set(b)))
```

```python
# 3. 리스트에서 중복되는 값만 set으로 변환해서 출력해주세요
nums = [1,2,2,3,4,4,4]

seen = set() # 처음 나온 값 저장
dup = set() # 이미 나온 값(seen에 이미 존재하는 값) 저장

for n in nums:
    if n in seen:
        dup.add(n)
    else:
        seen.add(n)    

print(dup)
```

```python
# 고객이 구매한 상품 종류
products = ["노트북", "마우스", "키보드", "마우스", "모니터", "키보드", "마우스"]

#1. products를 tuple로 변환하여 product_tuple에 저장
product_tuple = tuple(products)
print(product_tuple)
#2. 중복 상품을 제거하기 위해 set으로 변환
product_set = set(product_tuple)
print(product_set)
#3. 다시 list로 변환하여 product_list에 저장
product_list = list(product_set)
#4. 최종 상품 목록을 출력
print(product_list)
```
