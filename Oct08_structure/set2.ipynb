```python
# 리스트 -> set -> 리스트
user_ids= ['user1', 'user2','user1','user3','user3']
# 중복 제거
unique_user_ids = set(user_ids)
print(unique_user_ids)
# 인덱싱이 가능해지도록 list 변환 
unique_user_ids_list = list(unique_user_ids)
print(unique_user_ids_list)
print(unique_user_ids_list[0])
```

```python
# 한줄로 표현
print(list(set(user_ids)))
```

```python
# 집합 연산(두 데이터 간의 공통점, 차이점 비교)
python_class = {'scott','james','adam'}
js_class = {'bob', 'david', 'scott'}

# 합집합 union()
print('전체 학생1:', python_class.union(js_class))
print('전체 학생2:', python_class | js_class)

# 교집합 intersection()
print('둘 다 듣는 학생:', python_class.intersection(js_class))
print('둘 다 듣는 학생:', python_class & js_class)

# 차집함
print('파이썬만 듣는 학생:', python_class.difference(js_class))
print('JS만 듣는 학생:', js_class.difference(python_class))
print('파이썬만 듣는 학생1:', python_class-js_class)

```
