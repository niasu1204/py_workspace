```python
# id(변수) : 변수에 저장된 주소값 반환

s1 = 'ABC'
s2 = input('ABC입력해보기')

print(s1, id(s1))
print(s2, id(s2))
```

```python
s3 = 'ABC'

print(s1, id(s1))
print(s3, id(s3))#  새로 객체를 만들지 않고 주소값만 복사해서 사용
# 문자열 interning - 공백 없고 알파벳, 숫자, _ 로만 이루어진  20자 내외 문자열 대상
```

```python
# 프로그램 시작하면 
# heap 영역에 -5 ~ 256까지의 정수를 미리 메모리에 하나씩만 만들어두고 만들어져있는 객체 주소값 사용
# small integer caching
num1 = 10
num2 = 10

print(num1, id(num1))
print(num2, id(num2))

num1 = 1000
num2 = 1000

print(num1, id(num1))
print(num2, id(num2))
```
